```python
import numpy as np

# Data yang ingin diprediksi
data_uji = np.array([175, 67, 23, 97, 67, 79, 25, 15, 75, 95])

fitur = df.drop(columns="Status").values
df["Jarak"] = np.abs(fitur - data_uji).sum(axis=1)   # |x - y| dijumlahkan

df[["Status", "Jarak"]].sort_values("Jarak")
```

```python
def knn_manhattan(df, x_uji, k=4):
    X = df.drop(columns=["Status", "Jarak"], errors="ignore").values
    jarak = np.abs(X - np.array(x_uji)).sum(axis=1)
    idx = np.argsort(jarak)[:k]
    return df.iloc[idx]["Status"].value_counts().idxmax(), jarak[idx]

knn_manhattan(df, [175, 67, 23, 97, 67, 79, 25, 15, 75, 95], k=4)
```
